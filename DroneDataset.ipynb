```python
import glob, hashlib, os

seen = {}
removed = 0

for img in sorted(glob.glob('/content/YOLO_Drones/drone-dataset/**/images/*.*', recursive=True)):
    h = hashlib.md5(open(img, 'rb').read()).hexdigest()
    if h in seen:
        os.remove(img)
        lbl = img.replace('/images/', '/labels/').rsplit('.', 1)[0] + '.txt'
        if os.path.exists(lbl):
            os.remove(lbl)
        removed += 1
    else:
        seen[h] = img

print(f"Удалено дубликатов: {removed}")
```

```python
import glob, os, shutil
from sklearn.model_selection import GroupShuffleSplit

src_dir = '/content/YOLO_Drones/drone-dataset'
dst_dir = '/content/YOLO_Drones/drone-dataset-split'

# 1. Собираем все пары (картинка, разметка)
pairs = []
for img in sorted(glob.glob(f"{src_dir}/**/*.*", recursive=True)):
    if img.lower().endswith(('.jpg', '.jpeg', '.png')):
        lbl = img.replace('/images/', '/labels/').rsplit('.', 1)[0] + '.txt'
        if os.path.exists(lbl):
            pairs.append((img, lbl))

print(f"Найдено валидных пар (фото + разметка): {len(pairs)}")

# 2. Определяем группу для каждого файла по префиксу имени (например, video1_frame05 -> video1)
# Если файлы названы просто числами (0001.jpg, 0002.jpg), группируем блоками по 20 кадров подряд
groups = []
for idx, (img, _) in enumerate(pairs):
    name = os.path.splitext(os.path.basename(img))[0]
    prefix = name.rsplit('_', 1)[0] if '_' in name else f"block_{idx // 20}"
    groups.append(prefix)

# 3. Разбиваем по группам: 80% train, 15% val, 5% test (одно видео не попадёт в разные выборки)
gss1 = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
train_idx, temp_idx = next(gss1.split(pairs, groups=groups))

temp_pairs = [pairs[i] for i in temp_idx]
temp_groups = [groups[i] for i in temp_idx]

gss2 = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42) # 0.25 от 20% = 5% test
val_sub_idx, test_sub_idx = next(gss2.split(temp_pairs, groups=temp_groups))

splits = {
    'train': [pairs[i] for i in train_idx],
    'val':   [temp_pairs[i] for i in val_sub_idx],
    'test':  [temp_pairs[i] for i in test_sub_idx]
}

# 4. Копируем файлы в новую чистую структуру папок
for split_name, items in splits.items():
    os.makedirs(f"{dst_dir}/{split_name}/images", exist_ok=True)
    os.makedirs(f"{dst_dir}/{split_name}/labels", exist_ok=True)
    for img, lbl in items:
        shutil.copy2(img, f"{dst_dir}/{split_name}/images/{os.path.basename(img)}")
        shutil.copy2(lbl, f"{dst_dir}/{split_name}/labels/{os.path.basename(lbl)}")
    print(f"{split_name}: {len(items)} файлов")
```
